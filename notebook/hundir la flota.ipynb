```python
class Barco:
    def __init__(self, id: int, tipo: str, longitud: int, x: int, y: int, orientacion: str, vidas: int):
        self.id = id
        self.tipo = tipo
        self.longitud = longitud
        self.x = x
        self.y = y
        self.orientacion = orientacion
        self.vidas = vidas

    def asignar_posicion(self, x: int, y: int, orientacion: str):
        self.x = x
        self.y = y
        self.orientacion = orientacion

class Jugador:
    def __init__(self, nombre: str, barcos: list[Barco], es_maquina: bool = False):
        self.nombre = nombre
        self.barcos = barcos
        self.es_maquina = es_maquina
        self.tablero = Tablero.crear_tablero()
        self.mapa_barcos = {}
        self.barcos_hundidos = 0
        self.colocar_barcos_aleatoriamente()

    '''def colocar_barcos(self):
        for barco in self.barcos:
            #cont = 0
            x, y = barco.x, barco.y
            longitud = barco.longitud
            for _ in range(longitud):
                self.mapa_barcos[(x, y)] = barco

                if not self.es_maquina:
                    self.tablero[x][y] = 'B'
                if barco.orientacion == 'H':
                    y += 1
                else:
                    x += 1'''

    def colocar_barcos_aleatoriamente(self, dimension= 10, max_intentos= 1000):
        while True:
            self.tablero = Tablero.crear_tablero()
            self.mapa_barcos = {}
            ocupado = [[False for _ in range(dimension + 2)] for _ in range(dimension + 2)]
            barcos_ordenados = sorted(self.barcos, key=lambda b: b.longitud, reverse=True)
            hubo_fallo = False
            for barco in barcos_ordenados:
                colocado = False
                intentos = 0
                while not colocado and intentos < max_intentos:
                    barco.orientacion = random.choice(['H', 'V'])
                    longitud = barco.longitud
                    if barco.orientacion == 'H':
                        x = random.randint(0, dimension - 1)
                        y = random.randint(0, dimension - longitud)
                    else:
                        x = random.randint(0, dimension - longitud)
                        y = random.randint(0, dimension - 1)
                    puede_colocar = True
                    for i in range(longitud):
                        cx, cy = (x, y + i) if barco.orientacion == 'H' else (x + i, y)
                        for dx in [-1, 0, 1]:
                            for dy in [-1, 0, 1]:
                                nx, ny = cx + dx, cy + dy
                                if 0 <= nx < dimension and 0 <= ny < dimension:
                                    if ocupado[nx + 1][ny + 1]:
                                        puede_colocar = False
                                        break
                            if not puede_colocar:
                                break
                        if not puede_colocar:
                            break
                    if puede_colocar:
                        barco.asignar_posicion(x, y, barco.orientacion)
                        for i in range(longitud):
                            cx, cy = (x, y + i) if barco.orientacion == 'H' else (x + i, y)
                            self.mapa_barcos[(cx, cy)] = barco

                            if not self.es_maquina:
                                self.tablero[cx][cy] = 'B'

                            for dx in [-1, 0, 1]:
                                for dy in [-1, 0, 1]:
                                    nx, ny = cx + dx, cy + dy
                                    if 0 <= nx < dimension and 0 <= ny < dimension:
                                        ocupado[nx + 1][ny + 1] = True
                        colocado = True
                    intentos += 1
                # Si este barco no pudo colocarse, abortamos el for para reiniciar el tablero entero
                if not colocado:
                    hubo_fallo = True
                    break
            # Si ningún barco falló y tenemos las 20 casillas completas, terminamos con éxito
            if not hubo_fallo and len(self.mapa_barcos) == 20:
                break

    def disparar(self, rival: 'Jugador', coordenada_x: int = None, coordenada_y: int = None):
        if coordenada_x is None or coordenada_y is None:
            while True:
                coordenada_x = random.randint(0, 9)
                coordenada_y = random.randint(0, 9)

                if rival.tablero[coordenada_x][coordenada_y] not in ['D', 'X']:
                    break

        if (coordenada_x, coordenada_y) in rival.mapa_barcos:
            barco = rival.mapa_barcos[(coordenada_x, coordenada_y)]
            barco.vidas -= 1
            rival.tablero[coordenada_x][coordenada_y] = 'X'

            if barco.vidas == 0:
                self.barcos_hundidos += 1

            return coordenada_x, coordenada_y, barco.vidas, barco
        else:
            rival.tablero[coordenada_x][coordenada_y] = 'D'
            return coordenada_x, coordenada_y, -1, None

    '''def disparar(self, rival: 'Jugador'):
        bucle = True
        while bucle:
            disparo = True
            if not self.es_maquina:
                try:
                    coor_x = int(input("Introduzca la coordenada X a la que disparar:"))
                    coor_y = int(input("Introduzca la coordenada Y a la que disparar:"))
                except ValueError:
                    print("Error al introducir las coordenadas. Las coordenadas tienen que ser numeros entre 0 y 9")
                    continue

                if coor_x < 0 or coor_x > 9:
                    print("La coordenada X no puede ser menor a 0 ni mayor a 9")
                    disparo = False
                elif coor_y < 0 or coor_y > 9:
                    print("La coordenada Y no puede ser menor a 0 ni mayor a 9")
                    disparo = False
                elif rival.tablero[coor_x][coor_y] in ['D', 'X']:
                    disparo = False
                    print("Ya has disparado anteriormente a esa coordenada")
            else:
                coor_x = random.randint(0, 9)
                coor_y = random.randint(0, 9)
                #print("Dispara la maquina")
                if rival.tablero[coor_x][coor_y] in ['D', 'X']:
                    disparo = False  

            if disparo:
                vidas = 0
                bucle = False
                if (coor_x, coor_y) in rival.mapa_barcos:
                    barco = rival.mapa_barcos[(coor_x, coor_y)]
                    barco.vidas -= 1
                    vidas = barco.vidas
                    rival.tablero[coor_x][coor_y] = "X"
                else:
                    vidas = -1
                    rival.tablero[coor_x][coor_y] = "D"
                    
                print(f"X: {coor_x}\tY: {coor_y}")
                return vidas'''
                

class Tablero:
    def __init__(self, id: int, dimension: int, barcos):
            self.id_jugador = id
            self.dimension = dimension
            self.barcos = barcos

    @staticmethod
    def crear_tablero():
        tablero = []
        for i in range(10):
            linea = []
            for j in range(10):
                linea.append("A")
            tablero.append(linea)
        return tablero

    
```

```python
# Aqui va a ir la ventana de juego

listado_barcos = [
    Barco(barco["ID"], barco["Nombre"], barco["Longitud"], barco["X"], barco["Y"], barco["Orientacion"], barco["Vidas"]) 
    for barco in barcos
]

listado_barcos_maquina = [
    Barco(barco["ID"], barco["Nombre"], barco["Longitud"], 0, 0, 'H', barco["Vidas"]) 
    for barco in barcos_maquina
]
jugador = Jugador("Jugador 1", listado_barcos, False)
maquina = Jugador("Ordenador", listado_barcos_maquina, True)

# Ventana principal
ventana = Tk()
ventana.title("Hundir la Flota")
ventana.geometry("900x550")
ventana.config(bg="#1e293b")

# Barra superior donde aparecen los mensajes durante la partida
lbl_estado = Label(
    ventana,
    text="Empieza la partida",
    font=("Arial", 12, "bold"),
    bg="#334155",
    fg="#f8fafc",
    pady=10
)
lbl_estado.pack(fill="x", pady=(0, 10))

# Contenedor principal
contenedor = Frame(ventana, bg="#1e293b")
contenedor.pack(expand=True)

# Tablero izquierdo
frame_jugador = Frame(contenedor, bg="#0f172a", padx=10, pady=10, relief="ridge", bd=2)
frame_jugador.grid(row=0, column=0, padx=20)
Label(frame_jugador, text="TU FLOTA", font=("Arial", 11, "bold"), bg="#0f172a", fg="#38bdf8").grid(row=0, column=0, columnspan=10, pady=(0, 8))

# Tablero derecho
frame_maquina = Frame(contenedor, bg="#0f172a", padx=10, pady=10, relief="ridge", bd=2)
frame_maquina.grid(row=0, column=1, padx=20)
Label(frame_maquina, text="ENEMIGO", font=("Arial", 11, "bold"), bg="#0f172a", fg="#f43f5e").grid(row=0, column=0, columnspan=10, pady=(0, 8))

botones_maquina = {}

juego_activo = True
def turno_maquina():
    global juego_activo
    if not juego_activo:
        return
    mx, my, vidas, barco = maquina.disparar(jugador)
    btn = botones_jugador[(mx, my)]
    if vidas >= 0:
        btn.config(text="X", bg="#dc2626")  # Rojo fuego
        
        if vidas == 0:
            if maquina.barcos_hundidos == 10:
                juego_activo = False
                lbl_estado.config(text="¡Has perdido! Toda tu flota ha sido hundida", fg="#ef4444")
                return
            lbl_estado.config(text=f"La máquina ha hundido tu {barco.tipo}. La maquina vuelve a disparar", fg="#f87171")
        else:
            lbl_estado.config(text=f"Le han dado a tu {barco.tipo}. La maquina vuelve a disparar", fg="#fbbf24")
        ventana.after(700, turno_maquina)
    else:
        btn.config(text="D", bg="#0284c7")  # Azul
        lbl_estado.config(text="La máquina disparó al agua. Es tu turno", fg="#38bdf8")

def disparar_jugador(coor_x, coor_y):
    global juego_activo
    if not juego_activo:
        return
    btn = botones_maquina[(coor_x, coor_y)]
    btn.config(state="disabled")

    _, _, vidas, barco = jugador.disparar(maquina, coordenada_x=coor_x, coordenada_y=coor_y)
    if vidas >= 0:
        btn.config(text="X", bg="#dc2626", fg="white")
        
        if vidas == 0:
            if jugador.barcos_hundidos == 10:
                juego_activo = False
                lbl_estado.config(text="¡Victoria! Has hundido toda la flota enemiga.", fg="#4ade80")
                return
            lbl_estado.config(text=f"¡Hundido! Destruiste su {barco.tipo}. Repites disparo", fg="#4ade80")
        else:
            lbl_estado.config(text=f"Tocado en ({coor_x}, {coor_y}). Repites disparo", fg="#facc15")
        
    else:
        btn.config(text="D", bg="#38bdf8", fg="black")
        lbl_estado.config(text=f"Agua en ({coor_x}, {coor_y}). Turno de la máquina...", fg="#94a3b8")
        
        ventana.after(600, turno_maquina)

for x in range(10):
    for y in range(10):
        btn = Button(
            frame_maquina,
            text="A",
            width=2,
            height=1,
            font=("Arial", 9, "bold"),
            bg="#1e293b",
            fg="#64748b",
            activebackground="#38bdf8"
        )
        btn.config(command=lambda fila=x, col=y: disparar_jugador(fila, col))
        btn.grid(row=x + 1, column=y, padx=1, pady=1)

        botones_maquina[(x, y)] = btn
botones_jugador = {}

for x in range(10):
    for y in range(10):
        es_barco = (x, y) in jugador.mapa_barcos
        color_fondo = "#475569" if es_barco else "#0284c7" # Si es agua, azul, si es barco, gris
        texto = "B" if es_barco else "A"

        btn = Button(
            frame_jugador,
            text=texto,
            width=2,
            height=1,
            font=("Arial", 9, "bold"),
            bg=color_fondo,
            fg="white",
            state="disabled"
        )

        btn.grid(row=x + 1, column=y, padx=1, pady=1)

        botones_jugador[(x,y)] = btn

ventana.mainloop()
```
